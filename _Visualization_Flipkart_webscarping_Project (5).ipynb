```python
# Load your dataset
df = pd.read_csv("Laptop_data_50B.csv")

# Find categorical columns
categorical_cols = df.select_dtypes(include=['object', 'category']).columns
print("Categorical columns:", list(categorical_cols))
print("Number of categorical columns:", len(categorical_cols))

# Find numerical columns
numerical_cols = df.select_dtypes(include=['number']).columns
print("Numerical columns:", list(numerical_cols))
print("Number of numerical columns:", len(numerical_cols))
```

```python
continuous_cols = ['Price', 'RatingsCount', 'ReviewsCount']
df[continuous_cols].head()
```

```python
#numerical columns
#Central Tendency
for col in continuous_cols:
    print(f"\n{col} Central Tendency:")
    print(f"Mean: {df[col].mean()}")
    print(f"Median: {df[col].median()}")
    print(f"Mode: {df[col].mode()[0]}")
```

```python
#categorical columns
#Measures of Dispersion
for col in continuous_cols:
    print(f"\n{col} Dispersion:")
    print(f"Variance: {df[col].var()}")
    print(f"Standard Deviation: {df[col].std()}")
    print(f"Range: {df[col].max() - df[col].min()}")
    print(f"IQR: {df[col].quantile(0.75) - df[col].quantile(0.25)}")
```

```python
######Numerical Variabels
#a)Histogram/Distribution Plot
for col in continuous_cols:
    plt.figure(figsize=(8,4))
    sns.histplot(df[col], kde=True, bins=30)
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python
#b)Box Plot
for col in continuous_cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of{col}')
    plt.show()
```

```python
#Violin plot
for col in continuous_cols:
    plt.figure(figsize=(6,4))
    sns.violinplot(x=df[col])
    plt.title(f'Violin Plot of {col}')
    plt.show()
```

```python
#Scatter plot
num_cols = df.select_dtypes(include=['number'])
for i, col1 in enumerate(num_cols.columns):
    for j, col2 in enumerate(num_cols.columns):
        if i < j: 
            plt.figure(figsize=(6, 4))
            plt.scatter(df[col1], df[col2], alpha=0.6)
            plt.xlabel(col1)
            plt.ylabel(col2)
            plt.title(f'Scatter Plot: {col1} vs {col2}')
            plt.show()
```

```python
#Dist plot
# Select numerical columns
num_cols = df.select_dtypes(include=['number'])

# Plot dist plot for each numerical column
for col in num_cols.columns:
    plt.figure(figsize=(6, 4))
    sns.histplot(df[col], kde=True, bins=30)  # KDE curve + histogram
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

```python
#Pair plot
# Select numerical columns
num_cols = df.select_dtypes(include=['number'])

# Pair plot for all numerical columns
sns.pairplot(num_cols)
plt.show()
```

```python
####Categorical Variables
# Select categorical columns
cat_cols = df.select_dtypes(include=['object'])

# Loop through categorical columns and print frequency distribution
for col in cat_cols.columns:
    print(f"\n--- {col} ---")
    freq = df[col].value_counts()
    perc = df[col].value_counts(normalize=True) * 100
    freq_df = pd.DataFrame({
        'Count': freq,
        'Percentage': perc.round(2)
    })
    print(freq_df)
```

```python
# Count plot
# Select categorical columns
cat_cols = df.select_dtypes(include=['object'])

# Plot count plots for each categorical column
for col in cat_cols.columns:
    plt.figure(figsize=(8, 4))
    sns.countplot(x=col, data=df, order=df[col].value_counts().index)
    plt.xticks(rotation=45)
    plt.title(f'Count Plot of {col}')
    plt.show()
```

```python
#Bar plot 
# Select categorical columns
cat_cols = df.select_dtypes(include=['object'])

# Create bar plot for each categorical column
for col in cat_cols.columns:
    plt.figure(figsize=(8, 4))
    freq = df[col].value_counts()
    sns.barplot(x=freq.index, y=freq.values)
    plt.xticks(rotation=45)
    plt.title(f'Bar Plot of {col}')
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```

```python
#Pie chart 
# Select categorical columns
cat_cols = df.select_dtypes(include=['object'])

# Create pie chart for each categorical column
for col in cat_cols.columns:
    plt.figure(figsize=(5, 10))
    freq = df[col].value_counts()
    plt.pie(freq, labels=freq.index, autopct='%1.1f%%', startangle=90)
    plt.title(f'Pie Chart of {col}')
    plt.show()
```
