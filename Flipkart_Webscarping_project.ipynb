```python
LaptopNames = []
Price = []
Ratings = []
Processor = []
Page_num = []
for i in range(1,79):
    print('-------------------------------------------------------------------------------------------------')
    start  = time.time()
    URL='https://www.flipkart.com/search?q=laptops&sid=6bo%2Cb5g&as=on&as-show=on&otracker=AS_QueryStore_HistoryAutoSuggest_1_7_na_na_na&otracker1=AS_QueryStore_HistoryAutoSuggest_1_7_na_na_na&as-pos=1&as-type=HISTORY&suggestionId=laptops%7CLaptops&requestId=a66546a4-1232-4fc5-858b-8e3064926b0b&as-searchtext=Laptops'.format(i)
    page=requests.get(URL)
    pagecontent=page.text
    soup = BeautifulSoup(pagecontent)

    for x in soup.find_all('div',attrs={'class':'yKfJKb row'}):
        LaptopNames1 = x.find('div',attrs={'class':'KzDlHZ'})
        Price1 = x.find('div',attrs={'class':'Nx9bqj _4b5DiR'})
        Rating1 = x.find('span',attrs={'class':'Wphh3N'})
        Processor1 = x.find('li',attrs={'class':'J+igdf'})
        
        if LaptopNames1 is None:
            LaptopNames.append(np.nan)
        else:
            LaptopNames.append(LaptopNames1.text)

        if Price1 is None:
            Price.append(np.nan)
        else:
            Price.append(Price1.text)

        if Rating1 is None:
            Ratings.append(np.nan)   
        else:
            Ratings.append(Rating1.text)

        if Processor1 is None:
            Processor.append(np.nan)
        else:
            Processor.append(Processor1.text)
        Page_num.append(i)
    end = time.time()
    print('Page {} is completed in {} seconds'.format(i,end-start))   
```

```python
# Load the data from the uploaded CSV file.
df = pd.read_csv("Laptop_data_50B.csv")

###Removing Rupess Symbol
# Clean the 'Price' column: remove '₹' and commas, then convert to numeric
df['Price'] = df['Price'].str.replace('₹', '', regex=True)
df['Price'] = df['Price'].str.replace(',', '', regex=True)
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

# Save the cleaned DataFrame to a new CSV file
df.to_csv("Laptop_data_50B.csv", index=False)
```

```python
df = pd.read_csv("Laptop_data_50B.csv")

# Clean columns with list-like strings
def clean_string_column(series):
    if pd.api.types.is_object_dtype(series):
        series = series.str.replace("\[|\]|\'", "", regex=True)
        return series
    return series

for col in ['BrandName', 'ModelName', 'RAM', 'Storage']:
    df[col] = clean_string_column(df[col])

# Check for missing values in 'RAM' and 'Storage'
print("Missing values before imputation:")
print(df[['RAM', 'Storage']].isnull().sum())

# Find the mode (most frequent value) for 'RAM' and 'Storage'
ram_mode = df['RAM'].mode()[0]
storage_mode = df['Storage'].mode()[0]

print(f"\nMost frequent RAM value (mode): '{ram_mode}'")
print(f"Most frequent Storage value (mode): '{storage_mode}'")

# Impute missing values with the mode
df['RAM'].fillna(ram_mode, inplace=True)
df['Storage'].fillna(storage_mode, inplace=True)

# Check for missing values after imputation
print("\nMissing values after imputation:")
print(df[['RAM', 'Storage']].isnull().sum())

# Save the imputed data to a new CSV file
df.to_csv("Laptop_data_50B.csv", index=False)
```
